```python
accident_cols = [
    '2019 Accidents',
    '2020 Accidents',
    '2021 Accidents',
    '2022 Accidents',
    '2023 Accidents'
]

for col in accident_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df[accident_cols].dtypes
```

```python
forecast_df = df.copy()

forecast_df['Average'] = forecast_df[
    [
        '2019 Accidents',
        '2020 Accidents',
        '2021 Accidents',
        '2022 Accidents',
        '2023 Accidents'
    ]
].mean(axis=1)

forecast_df[['State', 'Average']].head()
```

```python
df.info()
```

```python
forecast_df['Risk Score'] = (
    forecast_df['2023 Accidents'] /
    forecast_df['2023 Accidents'].max()
) * 100

forecast_df[['State', 'Risk Score']].head()
```

```python
def risk_category(score):
    if score >= 75:
        return "High"
    elif score >= 40:
        return "Medium"
    else:
        return "Low"

forecast_df['Risk Category'] = forecast_df['Risk Score'].apply(risk_category)

forecast_df[['State', 'Risk Score', 'Risk Category']].head()
```

```python
dangerous = forecast_df.sort_values(
    by='Risk Score',
    ascending=False
)

dangerous[['State', 'Risk Score', 'Risk Category']].head(10)
```

```python
import matplotlib.pyplot as plt

top10 = dangerous.head(10)

plt.figure(figsize=(12,5))
plt.bar(top10['State'], top10['Risk Score'])

plt.xticks(rotation=45)
plt.title("Top 10 Dangerous States")
plt.ylabel("Risk Score")
plt.tight_layout()

plt.show()
```
